```python
bus_day["next_stop"] = bus_day["Stop ID"].shift(-1)
bus_day["next_sequence"] = bus_day["Stop Seq ID"].shift(-1)
bus_day["next_arrival"] = bus_day["arrival_time"].shift(-1)
bus_day["next_source_row"] = bus_day["source_row"].shift(-1)
bus_day["travel_time_s"] = (
    bus_day["next_arrival"] - bus_day["arrival_time"]
).dt.total_seconds()

pair_columns = ["source_row", "Stop ID", "next_source_row", "next_stop",
                "arrival_time", "next_arrival", "travel_time_s"]
bus_day[pair_columns].head(12)
```

```python
REVIEW_THRESHOLD_SECONDS = 600

sequence = pd.to_numeric(bus_day["Stop Seq ID"], errors="coerce")
ordinary_visit = (
    sequence.notna() & (sequence >= 0) & (sequence % 1 == 0) & (sequence != 999)
    & bus_day["Stop ID"].notna() & (bus_day["Stop ID"] != 9999)
)
bus_day["flag_sequence_or_id"] = ~(
    ordinary_visit & ordinary_visit.shift(-1, fill_value=False)
    & ((bus_day["next_sequence"] - sequence) == 1)
)
tied_time = bus_day["arrival_time"].duplicated(keep=False)
bus_day["flag_tied_time"] = tied_time | tied_time.shift(-1, fill_value=False)
bus_day["flag_missing_or_nonpositive"] = (
    bus_day["travel_time_s"].isna() | (bus_day["travel_time_s"] <= 0)
)
bus_day["flag_long_interval"] = bus_day["travel_time_s"] > REVIEW_THRESHOLD_SECONDS

flag_columns = ["flag_sequence_or_id", "flag_tied_time",
                "flag_missing_or_nonpositive", "flag_long_interval"]
bus_day["passes_basic_checks"] = ~bus_day[flag_columns].any(axis=1)
bus_day[pair_columns + flag_columns].head(12)
```

```python
accepted_indices = review_table.index[
    review_table["decision"].astype(str).str.strip().str.lower().eq("accept")
    & review_table["reason_or_evidence"].astype(str).str.strip().ne("")
]
current_accepted = pd.Series(bus_day.index.isin(accepted_indices), index=bus_day.index)
previous_accepted = current_accepted.shift(1, fill_value=False)

bus_day["previous_segment_time_s"] = bus_day["candidate_previous_time_s"].where(
    current_accepted & previous_accepted
)

feature_columns = ["source_row", "stop_pair", "hour", "weekday", "time_band",
                   "previous_segment_time_s", "travel_time_s"]
feature_examples = bus_day.loc[bus_day["passes_basic_checks"], feature_columns].copy()
feature_examples.head(8)
```

```python
raw = pd.read_excel("RTS_Route11_Spring2023.xlsx", engine="openpyxl")
display(raw.head())
```

```python
print("Actual-data shape:", raw.shape)
print("Buses:", raw["Bus"].nunique(), "Dates:", raw["Date"].nunique())
display(raw.isna().sum().to_frame("missing_values"))
```

```python
data = raw.copy()
data["source_row"] = data.index
data = data.drop_duplicates(subset=raw.columns.tolist()).copy()
data["bus_id"] = data["Bus"].astype("string")
data["stop_id"] = data["Stop ID"].astype("string").str.replace(r"\.0$", "", regex=True)
data["stop_sequence"] = pd.to_numeric(data["Stop Seq ID"], errors="coerce")
data["service_date"] = pd.to_datetime(data["Date"], errors="coerce").dt.normalize()
clock = pd.to_timedelta(data["Arrive"].astype(str), errors="coerce")
clock = clock.where((clock >= pd.Timedelta(0)) & (clock < pd.Timedelta(days=1)))
data["actual_arrival"] = data["service_date"] + clock

print("Rows before:", len(raw), "After exact-copy rule:", len(data))
display(data[["source_row", "bus_id", "stop_id", "stop_sequence", "actual_arrival"]].head())
```

```python
bad_fields = (
    data[["bus_id", "stop_id", "stop_sequence", "actual_arrival"]].isna().any(axis=1)
    | (data["stop_sequence"] < 0) | (data["stop_sequence"] % 1 != 0)
    | (data["stop_sequence"] == 999) | data["stop_id"].eq("9999").fillna(True)
)
repeated_key = data.duplicated(["bus_id", "service_date", "stop_id", "actual_arrival"], keep=False)
tied_time = data.duplicated(["bus_id", "service_date", "actual_arrival"], keep=False)
data["needs_review"] = bad_fields | repeated_key | tied_time

display(pd.Series({"invalid_fields": int(bad_fields.sum()),
                   "repeated_key_rows": int(repeated_key.sum()),
                   "tied_time_rows": int(tied_time.sum())}).to_frame("flagged_rows"))
display(data.loc[data["needs_review"], ["source_row", "bus_id", "stop_id", "actual_arrival"]].head(6))
```

```python
def read_schedule_table(filename):
    with zipfile.ZipFile("RTS_GTFS_Spring2023.zip") as archive:
        table = pd.read_csv(archive.open(filename), dtype=str, keep_default_na=False)
    table.columns = table.columns.str.strip()
    return table.apply(lambda column: column.str.strip())

trips = read_schedule_table("trips.txt")
stop_times = read_schedule_table("stop_times.txt")
calendar = read_schedule_table("calendar.txt")
exceptions = read_schedule_table("calendar_dates.txt")
stops = read_schedule_table("stops.txt")

route11_trips = trips.loc[trips["route_id"] == "11"]
display(route11_trips[["trip_id", "direction_id", "service_id", "trip_headsign"]].head())
```

```python
route11_times = route11_trips.merge(stop_times, on="trip_id", validate="one_to_many")
route11_times = route11_times.loc[route11_times["arrival_time"] != ""].copy()
route11_times = route11_times.merge(stops[["stop_id", "stop_name"]], on="stop_id", validate="many_to_one")

scheduled_days = []
for date in sorted(data["service_date"].dropna().unique()):
    date = pd.Timestamp(date)
    date_code = date.strftime("%Y%m%d")
    active = set(calendar.loc[
        (calendar["start_date"] <= date_code) & (calendar["end_date"] >= date_code)
        & (calendar[date.day_name().lower()] == "1"), "service_id"
    ])

    
    for change in exceptions.loc[exceptions["date"] == date_code].itertuples(index=False):
        if change.exception_type == "1":
            active.add(change.service_id)
        elif change.exception_type == "2":
            active.discard(change.service_id)
    day_schedule = route11_times.loc[route11_times["service_id"].isin(active)].copy()
    day_schedule["service_date"] = date
    day_schedule["scheduled_arrival"] = date + pd.to_timedelta(day_schedule["arrival_time"])
    scheduled_days.append(day_schedule)

schedule = pd.concat(scheduled_days, ignore_index=True)
schedule["stop_sequence"] = pd.to_numeric(schedule["stop_sequence"])
display(schedule[["service_date", "trip_id", "direction_id", "stop_id", "scheduled_arrival"]].head())
```

```python
pattern_check = schedule[["stop_id", "stop_name", "stop_sequence", "direction_id"]].drop_duplicates()
display(pattern_check.sort_values(["direction_id", "stop_sequence"]))
```

```python
observations = data.loc[~data["needs_review"]].copy()
possible = observations.merge(
    schedule, on=["service_date", "stop_id", "stop_sequence"], how="inner"
)
possible["matching_gap_minutes"] = (
    possible["actual_arrival"] - possible["scheduled_arrival"]
).dt.total_seconds().abs() / 60

MATCH_WINDOW_MINUTES = 20
possible = possible.loc[possible["matching_gap_minutes"] <= MATCH_WINDOW_MINUTES].copy()
possible["candidate_count"] = possible.groupby("source_row")["source_row"].transform("size")
matched = possible.loc[possible["candidate_count"] == 1].copy()

# Two observations must not silently claim the same scheduled stop visit.
conflict = matched.duplicated(["service_date", "trip_id", "stop_sequence"], keep=False)
conflict_review = matched.loc[conflict].copy()
matched = matched.loc[~conflict].copy()
matched["match_status"] = "provisional — requires journey review"

print("Unflagged recorded visits:", len(observations))
print("Visits with one non-conflicting proposed schedule match:", len(matched))
print("Conflicting assignments set aside:", len(conflict_review))
display(matched[["source_row", "bus_id", "stop_id", "direction_id", "trip_id",
                 "scheduled_arrival", "actual_arrival"]].head(10))
```

```python
matched["delay_minutes"] = (
    matched["actual_arrival"] - matched["scheduled_arrival"]
).dt.total_seconds() / 60
display(matched[["stop_id", "scheduled_arrival", "actual_arrival", "delay_minutes"]].head())
```

```python
def timepoint_anchors(block,target_sequence):
    """Other unique recorded timepoint positions, retaining ambiguity flags."""
    keys=planned[['stop_id','stop_sequence']].drop_duplicates()
    anchors=block.merge(keys,on=['stop_id','stop_sequence'],how='inner')
    anchors=anchors[anchors.stop_sequence.ne(target_sequence)&anchors.actual_arrival.notna()].copy()
    valid=anchors.groupby('stop_sequence').actual_arrival.transform('nunique').eq(1)
    return anchors[valid].drop_duplicates(['stop_sequence','stop_id','actual_arrival'])

def external_candidate_ids(anchors,observed_pattern,day):
    """Leave target arrival entirely out of the assignment check."""
    if anchors.empty or len(observed_pattern)<5:return []
    candidate_ids=[]
    for tid,t in date_trips[day].groupby('trip_id'):
        expected=dict(full_patterns[tid])
        if any(expected.get(seq)!=stop for seq,stop in observed_pattern):continue
        joined=anchors.merge(t[['stop_id','stop_sequence','scheduled_arrival']],on=['stop_id','stop_sequence'])
        if joined.stop_sequence.nunique()!=anchors.stop_sequence.nunique():continue
        offset=(joined.actual_arrival-joined.scheduled_arrival).dt.total_seconds()/60
        if offset.abs().le(20).all():candidate_ids.append(tid)
    return candidate_ids
```

```python
audit_rows=[]
for visit in matched.itertuples():
    block=blocks[block_by_source.loc[visit.source_row]]
    ordinary=block[block.stop_sequence.between(0,32)&block.stop_id.ne('9999')]
    observed=list(dict.fromkeys(zip(ordinary.stop_sequence,ordinary.stop_id)))
    expected=full_patterns[visit.trip_id]
    expected_map=dict(expected)
    other_pattern=[(seq,stop) for seq,stop in observed if seq!=visit.stop_sequence]
    all_anchors=timepoint_anchors(block,visit.stop_sequence)
    strict=all_anchors[~all_anchors.needs_review]
    weak_ids=external_candidate_ids(all_anchors,other_pattern,visit.service_date)
    strict_ids=external_candidate_ids(strict,other_pattern,visit.service_date)
    if not all_anchors.empty and len(other_pattern)>=5:
        if len(weak_ids)==1:
            status='same_trip_supported_by_other_anchors' if weak_ids[0]==visit.trip_id else 'different_trip_suggested'
        elif len(weak_ids)>1:status='ambiguous_other_anchors'
        else:status='other_anchors_do_not_support_current_trip'
    else:status='insufficient_external_anchor_or_pattern_evidence'
    audit_rows.append({'source_row':visit.source_row,'date':str(visit.service_date.date()),'bus_id':visit.bus_id,
        'stop_id':visit.stop_id,'stop_sequence':visit.stop_sequence,'direction_id':visit.direction_id,
        'current_trip_id':visit.trip_id,'block_id':block_by_source.loc[visit.source_row],
        'full_ordered_pattern_matches':observed==expected,
        'pattern_mismatches':sum(expected_map.get(seq)!=stop for seq,stop in observed),
        'source_times_monotonic':block.actual_arrival.is_monotonic_increasing,
        'other_unique_timed_anchors':len(all_anchors),'other_unflagged_timed_anchors':len(strict),
        'external_candidate_count':len(weak_ids),'external_candidate_trip_ids':';'.join(weak_ids),
        'strict_external_candidate_count':len(strict_ids),
        'strict_same_trip_supported':len(strict_ids)==1 and strict_ids[0]==visit.trip_id,
        'external_check_status':status,
        'source_block_rows':','.join(str(int(x)) for x in block.source_row)})
linkage_audit=pd.DataFrame(audit_rows)
linkage_audit['recommended_review_action']='hold_terminal_dependent_corroboration'
linkage_audit.loc[linkage_audit.strict_same_trip_supported,'recommended_review_action']='priority_review_unflagged_anchor_support'
linkage_audit.loc[linkage_audit.stop_id.isin(['1','326']),'recommended_review_action']='hold_terminal_timestamp_semantics'
linkage_audit.loc[linkage_audit.external_check_status.ne('same_trip_supported_by_other_anchors'),'recommended_review_action']='hold_insufficient_independent_evidence'
matched_with_blocks=matched.copy()
matched_with_blocks['block_id']=matched_with_blocks.source_row.map(block_by_source)
block_trip_counts=matched_with_blocks.groupby('block_id').trip_id.nunique()
chronology=matched_with_blocks.sort_values(['bus_id','service_date','source_row'])
schedule_steps=chronology.groupby(['bus_id','service_date']).scheduled_arrival.diff()
chronology_inversions=int(schedule_steps.lt(pd.Timedelta(0)).sum())
```

```python

# Independent source-order evidence for terminal timestamp interpretation.

dev['next_stop_id'] = dev.groupby(['bus_id','service_date']).stop_id.shift(-1)
dev['next_sequence'] = dev.groupby(['bus_id','service_date']).stop_sequence.shift(-1)
dev['next_arrival'] = dev.groupby(['bus_id','service_date']).actual_arrival.shift(-1)

# Route 11 terminal is sequence 32, followed by sequence 0.
terminal_transition = dev.stop_sequence.eq(32) & dev.next_sequence.eq(0)

terminal_examples = dev.loc[
    terminal_transition,
    ['source_row','bus_id','service_date','stop_id',
     'actual_arrival','next_stop_id','next_arrival']
].copy()

terminal_examples['same_stop'] = terminal_examples.stop_id.eq(
    terminal_examples.next_stop_id
)
terminal_examples['same_timestamp'] = terminal_examples.actual_arrival.eq(
    terminal_examples.next_arrival
)

# Generate destination-arrival review candidates.
endpoint_review_rows = []

for block_id, block in blocks.items():

    ordinary = block[
        block.stop_sequence.between(0,32) &
        block.stop_id.ne('9999')
    ]

    observed = list(dict.fromkeys(
        zip(ordinary.stop_sequence, ordinary.stop_id)
    ))

    complete_ids = [
        tid for tid in date_trips[block.service_date.iloc[0]].trip_id.unique()
        if observed == full_patterns[tid]
    ]

    if not complete_ids or not block.actual_arrival.is_monotonic_increasing:
        continue

    ends = ordinary[ordinary.stop_sequence.eq(32)]

    if ends.empty or ends.actual_arrival.nunique() != 1:
        continue

    endpoint = ends.iloc[0]

    anchors = timepoint_anchors(block,32)
    anchors = anchors[
        anchors.stop_id.eq('107') & ~anchors.needs_review
    ]

    ids = external_candidate_ids(
        anchors,
        [(seq,stop) for seq,stop in observed if seq != 32],
        endpoint.service_date
    )

    if len(ids) != 1:
        continue

    tid = ids[0]

    planned_end = planned[
        (planned.service_date == endpoint.service_date) &
        planned.trip_id.eq(tid) &
        planned.stop_sequence.eq(32)
    ].iloc[0]

    offset = (
        endpoint.actual_arrival - planned_end.scheduled_arrival
    ).total_seconds()/60

    endpoint_review_rows.append({
        'source_row': int(endpoint.source_row),
        'all_source_rows': ','.join(str(int(x)) for x in ends.source_row),
        'date': str(endpoint.service_date.date()),
        'bus_id': endpoint.bus_id,
        'block_id': block_id,
        'stop_id': endpoint.stop_id,
        'stop_sequence': 32,
        'trip_id': tid,
        'direction_id': planned_end.direction_id,
        'scheduled_arrival': planned_end.scheduled_arrival,
        'recorded_arrival': endpoint.actual_arrival,
        'offset_minutes_for_review_only': offset,
        'outside_original_20min_window': abs(offset)>20,
        'linking_anchor_source_rows': ','.join(
            str(int(x)) for x in anchors.source_row
        ),
        'was_in_current_modeling_table':
            endpoint.source_row in set(matched.source_row),
        'status': 'destination_arrival_candidate_requires_manual_review'
    })

endpoint_review = pd.DataFrame(endpoint_review_rows)
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.pipeline import Pipeline
from sklearn.linear_model import LinearRegression, Ridge
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

# Select only information available before the arrival occurs.
features = ["direction_id", "scheduled_hour", "weekday"]
target = "delay_minutes"

train = cleaned_arrivals.loc[
    cleaned_arrivals["split"].eq("train")
].copy()

validation = cleaned_arrivals.loc[
    cleaned_arrivals["split"].eq("validation")
].copy()

assert not train.empty
assert not validation.empty
assert train["service_date"].max() < validation["service_date"].min()
assert "bus_id" not in features

print("Training observations:", len(train))
print("Validation observations:", len(validation))

display(train[features + [target]].head())
display(
    cleaned_arrivals.groupby(
        ["split", "service_date", "bus_id"]
    ).size().rename("records").reset_index()
)

# Prepare the regression models.
def make_model(name):
    preparation = ColumnTransformer([
        (
            "categories",
            OneHotEncoder(
                handle_unknown="ignore",
                drop="first",
                sparse_output=False
            ),
            ["direction_id", "weekday"]
        ),
        ("hour", StandardScaler(), ["scheduled_hour"])
    ])

    estimator = (
        LinearRegression()
        if name == "Linear regression"
        else Ridge(alpha=100)
    )

    return Pipeline([
        ("prepare", preparation),
        ("model", estimator)
    ])


def predict_method(name, training, evaluation):
    if name == "Published schedule":
        return np.zeros(len(evaluation))

    if name == "Training median":
        return np.repeat(training[target].median(), len(evaluation))

    fitted = make_model(name).fit(
        training[features],
        training[target]
    )

    return fitted.predict(evaluation[features])


methods = [
    "Published schedule",
    "Training median",
    "Linear regression",
    "Ridge regression"
]

# Expanding whole-date cross-validation using training dates only.
dates = sorted(train["service_date"].unique())

fold_rows = []
fold_predictions = []

for fold, end in enumerate([4, 6, 8], start=1):
    fold_train = train.loc[train["service_date"].isin(dates[:end])]
    fold_valid = train.loc[train["service_date"].isin(dates[end:end+2])]

    if fold_train.empty or fold_valid.empty:
        raise ValueError("A cross-validation fold has no observations.")

    assert fold_train["service_date"].max() < fold_valid["service_date"].min()

    for name in methods:
        prediction = predict_method(name, fold_train, fold_valid)
        errors = np.abs(fold_valid[target].to_numpy() - prediction)

        fold_rows.append({
            "fold": fold,
            "method": name,
            "training_rows": len(fold_train),
            "validation_rows": len(fold_valid),
            "MAE_minutes": errors.mean(),
            "train_through": fold_train["service_date"].max(),
            "validate_from": fold_valid["service_date"].min()
        })

        part = fold_valid[
            ["source_row", "service_date", target]
        ].copy()

        part["method"] = name
        part["absolute_error_minutes"] = errors
        fold_predictions.append(part)

fold_scores = pd.DataFrame(fold_rows)
oof_predictions = pd.concat(fold_predictions, ignore_index=True)

cv_summary = (
    oof_predictions.groupby("method")["absolute_error_minutes"]
    .agg(["count", "mean"])
    .rename(columns={"mean": "MAE_minutes"})
    .sort_values("MAE_minutes")
)

selected_method = cv_summary.index[0]

display(fold_scores)
display(cv_summary)

print("Selected using training-date folds:", selected_method)

# Evaluate all methods on the reserved validation dates.
score_rows = []

validation_predictions = validation[
    ["source_row", "bus_id", "service_date",
     "direction_id", "scheduled_hour", target]
].copy()

for name in methods:
    prediction = predict_method(name, train, validation)
    validation_predictions[name] = prediction

    score_rows.append({
        "method": name,
        "selected_by_training_folds": name == selected_method,
        "MAE_minutes": mean_absolute_error(validation[target], prediction),
        "RMSE_minutes": np.sqrt(
            mean_squared_error(validation[target], prediction)
        ),
        "R2": r2_score(validation[target], prediction)
    })

validation_scores = pd.DataFrame(score_rows)

selected_prediction = validation_predictions[selected_method]

validation_predictions["prediction_minutes"] = selected_prediction
validation_predictions["signed_error_minutes"] = (
    selected_prediction - validation[target]
)
validation_predictions["absolute_error_minutes"] = (
    validation_predictions["signed_error_minutes"].abs()
)

display(validation_scores.round(4))
display(validation_predictions.head(10))
```

```python

import statsmodels.api as sm
from itertools import combinations
from sklearn.ensemble import RandomForestRegressor
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder
from sklearn.pipeline import Pipeline
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

# Use the Route 11 data already created in Notebook 3.
arrivals = cleaned_arrivals.copy()
train = arrivals.loc[arrivals["split"].eq("train")].copy()
validation = arrivals.loc[arrivals["split"].eq("validation")].copy()

GROUPS = ["direction_id", "scheduled_hour", "weekday"]
TARGET = "delay_minutes"

# Check the data before fitting the model.
assert not train.empty and not validation.empty
assert train["service_date"].max() < validation["service_date"].min()
assert arrivals["source_row"].is_unique
assert not arrivals[[*GROUPS, TARGET, "service_date"]].isna().any().any()

assert np.allclose(
    arrivals[TARGET],
    (arrivals["actual_arrival"] - arrivals["scheduled_arrival"])
    .dt.total_seconds() / 60
)

print("Training:", len(train), "records on", train.service_date.nunique(), "days")
print("Validation:", len(validation), "records on", validation.service_date.nunique())

display(arrivals[["service_date", "bus_id", *GROUPS, TARGET]].head())


# Create consistent indicator columns for categorical predictors.
def design(training, evaluation, groups):
    if not groups:
        return (
            pd.DataFrame({"const": 1.0}, index=training.index),
            pd.DataFrame({"const": 1.0}, index=evaluation.index)
        )

    a = training[list(groups)].copy()
    b = evaluation[list(groups)].copy()

    categories = [
        g for g in groups if g in ["direction_id", "weekday"]
    ]

    for g in categories:
        levels = sorted(a[g].dropna().unique())
        a[g] = pd.Categorical(a[g], categories=levels)
        b[g] = pd.Categorical(b[g], categories=levels)

    a = pd.get_dummies(
        a, columns=categories, drop_first=True, dtype=float
    )

    b = pd.get_dummies(
        b, columns=categories, drop_first=True, dtype=float
    )

    b = b.reindex(columns=a.columns, fill_value=0)

    return (
        sm.add_constant(a.astype(float), has_constant="add"),
        sm.add_constant(b.astype(float), has_constant="add")
    )


def fit_ols(training, groups):
    x, _ = design(training, training, groups)
    return sm.OLS(training[TARGET].astype(float), x).fit()


def ols_predict(training, evaluation, groups):
    fitted = fit_ols(training, groups)
    _, x = design(training, evaluation, groups)
    return fitted.predict(x).to_numpy()


# Fit the full regression model.
full_model = fit_ols(train, GROUPS)

print(full_model.summary())

# Account for repeated observations within the same service date.
clustered = full_model.get_robustcov_results(
    cov_type="cluster",
    groups=train.service_date,
    use_correction=True
)

uncertainty = pd.DataFrame({
    "coefficient_minutes": full_model.params,
    "ordinary_p_value": full_model.pvalues,
    "day_clustered_standard_error": clustered.bse,
    "day_clustered_p_value": clustered.pvalues
})

display(uncertainty.round(4))

print(
    "Training design rank:",
    np.linalg.matrix_rank(full_model.model.exog),
    "/",
    full_model.model.exog.shape[1]
)
```

```python

# Create expanding whole-date folds using training data only.
training_dates = sorted(train.service_date.unique())

folds = []

for end in [4, 6, 8]:
    if end + 2 > len(training_dates):
        continue

    a = train.loc[
        train.service_date.isin(training_dates[:end])
    ].copy()

    b = train.loc[
        train.service_date.isin(training_dates[end:end+2])
    ].copy()

    if a.empty or b.empty:
        continue

    assert a.service_date.max() < b.service_date.min()
    folds.append((a, b))

if not folds:
    raise ValueError("Not enough training dates to create the model folds.")


# Calculate average prediction error across the folds.
def fold_mae(groups):
    errors = []

    for a, b in folds:
        errors.extend(
            np.abs(b[TARGET].to_numpy() - ols_predict(a, b, groups))
        )

    return float(np.mean(errors))


# Compare all combinations of predictor groups.
subsets = [
    list(c)
    for size in range(len(GROUPS) + 1)
    for c in combinations(GROUPS, size)
]

rows = []

for groups in subsets:
    fitted = fit_ols(train, groups)

    rows.append({
        "predictors": " + ".join(groups) or "Intercept only",
        "parameters": len(fitted.params),
        "training_R2": fitted.rsquared,
        "adjusted_R2": fitted.rsquared_adj,
        "F_statistic": fitted.fvalue,
        "F_p_value": fitted.f_pvalue,
        "AIC": fitted.aic,
        "BIC": fitted.bic,
        "training_fold_MAE_minutes": fold_mae(groups)
    })

subset_results = pd.DataFrame(rows)

display(subset_results.round(4))


# Backward selection using AIC.
def backward_aic(training):
    kept = list(GROUPS)
    path = []
    candidates = []
    step = 0

    while True:
        current = fit_ols(training, kept)

        path.append({
            "step": step,
            "predictors": " + ".join(kept) or "Intercept only",
            "AIC": current.aic,
            "BIC": current.bic
        })

        options = []

        for dropped in kept:
            trial = [g for g in kept if g != dropped]
            fitted = fit_ols(training, trial)

            options.append((fitted.aic, dropped, trial))

            candidates.append({
                "step": step,
                "remove": dropped,
                "remaining": " + ".join(trial) or "Intercept only",
                "AIC": fitted.aic
            })

        if not options:
            break

        best_aic, dropped, trial = min(
            options, key=lambda item: item[0]
        )

        if best_aic >= current.aic - 1e-8:
            break

        kept = trial
        step += 1

    return kept, pd.DataFrame(path), pd.DataFrame(candidates)


selected_groups, selection_path, removal_candidates = backward_aic(train)

display(removal_candidates.round(3))
display(selection_path.round(3))

selected_model = fit_ols(train, selected_groups)

print("Predictor groups retained:", selected_groups)
print(selected_model.summary())


# Build the random forest model.
def forest():
    encoder = ColumnTransformer([
        (
            "categories",
            OneHotEncoder(
                handle_unknown="ignore",
                sparse_output=False
            ),
            ["direction_id", "weekday"]
        ),
        ("hour", "passthrough", ["scheduled_hour"])
    ])

    return Pipeline([
        ("encode", encoder),
        ("model", RandomForestRegressor(
            n_estimators=300,
            max_depth=4,
            min_samples_leaf=10,
            random_state=42,
            n_jobs=1
        ))
    ])


# Compare all five modeling procedures.
def predict_procedure(name, training, evaluation):
    if name == "Published schedule":
        return np.zeros(len(evaluation)), "none"

    if name == "Historical median":
        return (
            np.repeat(training[TARGET].median(), len(evaluation)),
            "none"
        )

    if name == "Full linear regression":
        return ols_predict(training, evaluation, GROUPS), ", ".join(GROUPS)

    if name == "Backward AIC regression":
        groups, _, _ = backward_aic(training)
        return (
            ols_predict(training, evaluation, groups),
            ", ".join(groups) or "intercept"
        )

    fitted = forest().fit(training[GROUPS], training[TARGET])

    return fitted.predict(evaluation[GROUPS]), ", ".join(GROUPS)


METHODS = [
    "Published schedule",
    "Historical median",
    "Full linear regression",
    "Backward AIC regression",
    "Random forest"
]

fold_scores = []
fold_predictions = []

for fold_number, (a, b) in enumerate(folds, start=1):

    for name in METHODS:
        predicted, used = predict_procedure(name, a, b)

        errors = b[TARGET].to_numpy() - predicted

        fold_scores.append({
            "fold": fold_number,
            "method": name,
            "training_rows": len(a),
            "evaluation_rows": len(b),
            "predictors": used,
            "MAE_minutes": np.abs(errors).mean()
        })

        for row_id, error in zip(b.source_row, errors):
            fold_predictions.append({
                "source_row": row_id,
                "method": name,
                "absolute_error": abs(error),
                "squared_error": error**2
            })


fold_scores = pd.DataFrame(fold_scores)
oof = pd.DataFrame(fold_predictions)

procedure_scores = oof.groupby("method").agg(
    records=("absolute_error", "size"),
    MAE_minutes=("absolute_error", "mean"),
    MSE=("squared_error", "mean")
)

procedure_scores["RMSE_minutes"] = np.sqrt(
    procedure_scores.pop("MSE")
)

procedure_scores = procedure_scores.sort_values("MAE_minutes")

chosen_procedure = procedure_scores.index[0]

display(fold_scores.round(3))
display(procedure_scores.round(4))

print("Procedure chosen using training folds only:", chosen_procedure)
```
